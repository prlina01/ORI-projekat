```python
# Cell 2: imports (minimalno)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
plt.style.use('seaborn-whitegrid')
%matplotlib inline
```

## Opis skupa podataka
- Izvor: ručno prikupljeni oglasi polovnih automobila dostupnih u Srbiji.
- Veličina: 1.756 zapisa, 10 ulaznih atributa + ciljna vrednost `Cena` (EUR).
- Pojašnjenje kolona:
  - **Marka**, **Grad** (kategorije) — segmentacija tržišta.
  - **Cena** – cilj predikcije.
  - **Godina_proizvodnje**, **Kilometraza**, **Konjske_snage**, **Zapremina_motora** – numerički atributi.
  - **Karoserija**, **Gorivo**, **Menjac** – kategorije koje utiču na vrednost vozila.

U narednim ćelijama kratko pregledamo sirove podatke i pripremamo dodatne kolone (`Log_Cena`, `Starost`) za analizu i modelovanje.

```python
# Cell 4: load and preview
data_path = Path('kola_skup_podataka.tsv')
assert data_path.exists(), 'kola_skup_podataka.tsv nije pronađen u radnom direktorijumu'
df = pd.read_csv(data_path, sep='	')
# Uredniji nazivi kolona (kratko)
df = df.rename(columns={
    'Godina proizvodnje': 'Godina_proizvodnje',
    'Zapremina motora': 'Zapremina_motora',
    'Konjske snage': 'Konjske_snage'
})
df['Log_Cena'] = np.log1p(df['Cena'])
current_year = pd.Timestamp.now().year
df['Starost'] = current_year - pd.to_numeric(df['Godina_proizvodnje'], errors='coerce').fillna(current_year).astype(int)
print('Broj redova, kolona:', df.shape)
df.head(6)
```

**Napomena za prezentaciju:** Obratiti pažnju na raznovrsnost gradova i raspon godina proizvodnje. Dodata kolona `Starost` (trenutna godina – godina proizvodnje) koristi se kasnije u LOESS analizi i feature engineering-u.

### Osnovne statistike i kvalitet podataka
Pre nego što pređemo na grafike, beležimo osnovne metrike koje se mogu citirati tokom izlaganja: medijana cene, tipična kilometraža, kao i informacija da u skupu nema nedostajućih vrednosti.

```python
# Statistički pregled (medijana, kvartili, kvalitet)
numeric_cols = ['Cena', 'Kilometraza', 'Konjske_snage', 'Zapremina_motora', 'Starost']
summary = df[numeric_cols].describe(percentiles=[0.25, 0.5, 0.75]).T
summary[['mean', '50%']] = summary[['mean', '50%']].round(1)
summary[['std', 'min', 'max']] = summary[['std', 'min', 'max']].round(1)
display(summary[["count", "mean", "50%", "std", "min", "max"]])
missing = df.isna().sum()
display(pd.DataFrame({'Broj nedostajućih vrednosti': missing[missing > 0]}))
if missing.sum() == 0:
    print('Dataset nema nedostajućih vrednosti.')
```

**Zaključci iz tabele:** Medijana cene je oko 7.500 EUR, dok gornji kvartil prelazi 14.000 EUR. Kilometraža je visoko varijabilna (std > 90.000 km), što potvrđuje potrebu za robustnim skaliranjem. U datasetu nema praznih vrednosti, ali kod koji je razvijen ipak ima fallback mehanizme za generisanje modela na budućim verzijama skupa.

## Istraživačka analiza podataka (EDA)
Ovom sekcijom odgovaramo na pitanja:
1. Kako izgleda raspodela cena i da li postoji jaka asimetrija koja otežava modelovanje?
2. Kakav je odnos između kilometraže i (log) cene — potvrđuje li intuiciju o padu vrednosti?
3. Koje marke dominiraju na tržištu i zahtevaju posebnu pažnju pri enkodiranju kategorija?

Grafici u nastavku služe kao materijal za komentarisanje tokom prezentacije (slike možete preneti i na slajdove).

```python
# Cell 6: EDA plots (concise)
fig, axes = plt.subplots(1, 3, figsize=(18,4))
sns.histplot(df['Cena'], bins=40, ax=axes[0], kde=False)
axes[0].set_title('Raspodela Cena (EUR)')
sns.histplot(df['Log_Cena'], bins=40, ax=axes[1], color='C1')
axes[1].set_title('Raspodela log1p(Cena)')
sns.scatterplot(x='Kilometraza', y='Log_Cena', data=df.sample(frac=0.25, random_state=1), ax=axes[2], alpha=0.6)
axes[2].set_title('Kilometraza vs log cena (sample)')
plt.tight_layout()
plt.show()

# Top 8 marki po broju oglasa
top_brands = df['Marka'].value_counts().nlargest(8)
plt.figure(figsize=(8,4))
sns.barplot(x=top_brands.values, y=top_brands.index, palette='muted')
plt.title('Top 8 brandova po broju oglasa')
plt.xlabel('Broj oglasa')
plt.show()
```

### Kratki zaključci iz EDA
- Raspodela cene je izrazito desno nagnuta, dok log-transformacija (`log1p`) dovodi distribuciju bliže normalnoj – neophodno za modele koji pretpostavljaju simetričnije greške.
- Negativna korelacija između kilometraže i log-cene potvrđuje očekivano ponašanje (više kilometara → niža cena), ali su prisutni outlieri koje ćemo tretirati IsolationForest-om i LOESS-om.
- Osam najzastupljenijih marki čini preko 50% skupa — u preprocesiranju obezbeđujemo kvalitetno enkodiranje kategorija (entity embeddings za TabTransformer, OneHot za GBR).

## Preprocesiranje i inženjering atributa
Pre nego što treniramo modele, u `main.py` primenjujemo sledeće korake (vredni pomena tokom prezentacije):
1. **Popunjavanje vrednosti** – iako u trening skupu nema praznina, naš kod predviđa situacije u kojima fali neka vrednost: medijana za numeričke, mod za kategorije.
2. **Detekcija i tretman outliera**:
   - *IsolationForest* (kontaminacija 2%) uklanja ekstremne kombinacije cene/kilometraže/snaga koji remete modele.
   - *LOESS soft clipping* po starosti vozila ublažava preostale ekstremne vrednosti, ali ih ne izbacuje (zadržavamo informaciju, smanjujemo težinu).
3. **Feature engineering**:
   - `Starost = Trenutna_godina – Godina_proizvodnje`.
   - `Deprecijacioni_indeks = (Procena_nove_cene – Trenutna_cena) / Starost^{α}`, gde α zavisi od marke (skuplji brendovi sporije gube vrednost).
   - Geo feature `Region` dobijen mapiranjem grada u 5 regiona Srbije (Beograd, Vojvodina, Zapadna, Istočna, Južna).
4. **Skaliranje** – koristimo `RobustScaler` nad numeričkim kolonama (medijana + IQR, otpornije na outliere).
5. **Enkodiranje kategorija**:
   - `GradientBoostingRegressor`: klasični `OneHotEncoder`.
   - `TabTransformer`: svaki kategorijski atribut ulazi u embedding sloj (dim 32), zatim kroz 4 transformer bloka.
6. **Evaluacija** – umesto proste 80/20 podele, koristimo *spatial-temporal* kros-validaciju: svaki region je jednom u validacionom fold-u + "mladi" automobili (posle 2018.) iz trening regiona.

### Spatial-temporal cross-validation (5 fold)
- Svaki fold izostavlja **ceo region** (npr. Vojvodina) iz treninga i koristi ga za validaciju.
- Pored toga, u valu se nalaze i novija vozila (godina > 2018) iz ostalih regiona kako bi model bio robustan na nove modele automobila.
- Ova strategija simulira realan scenario: procenjujemo cenu automobila u regionu gde model dosad nije treniran i za automobile koji su novijeg datuma.
- Ukupno trajanje treninga je duže nego kod standardne 80/20 podele, ali dobijamo kredibilniju metodu evaluacije i realističniju metriku (RMSE).

### Kratka arhitektura modela
- **GradientBoostingRegressor**: ansambl 500–800 plitkih stabala, uči rezidualne greške; brz i interpretabilan.
- **TabTransformer**:
  - Svaku kategoriju mapira u embedding vektor dimenzije 32.
  - 4 transformer bloka sa multi-head self-attention mehanizmom (8 glava).
  - Kombinuje izlaznu reprezentaciju sa numeričkim atributima kroz MLP koji daje predikciju log-cene.
- Oba modela treniraju nad log-transformisanom cenom; završna RMSE metrika se vraća u originalnu skalu (`exp1m`).

```python
# Cell 9: Pokreni main.py (skripta radi quick run automatski)
import subprocess, sys, re, pandas as _pd
cmd = [sys.executable, 'main.py', str(data_path)]
print('Pokrećem komandu:', ' '.join(cmd))
proc = subprocess.run(cmd, capture_output=True, text=True)
print('--- STDOUT ---')
print(proc.stdout)
print('--- STDERR (ukoliko postoji) ---')
print(proc.stderr if proc.stderr.strip() else 'nema grešaka')
rmse_pattern = re.compile(r"(GradientBoostingRegressor|TabTransformer) RMSE: ([0-9.]+)")
records = []
for line in proc.stdout.splitlines():
    match = rmse_pattern.search(line)
    if match:
        model_name, value = match.groups()
        records.append({'Model': model_name, 'RMSE (EUR)': float(value)})
if records:
    display(_pd.DataFrame(records))
else:
    print('Nije pronađen RMSE u izlazu – proveriti logove iznad.')
```

## Rezultati modela iz `main.py`
Skripta `main.py` sada automatski koristi "quick" podešavanje (optimizovani parametri, kraći trening) i vraća RMSE za obe arhitekture. Tokom prezentacije pokrenite ćeliju uživo da pokažete kako se rezultati generišu i da je kod reproduktivan.

### Interpretacija rezultata
- `GradientBoostingRegressor` i dalje pokazuje stabilne rezultate (~4k EUR RMSE u quick modu).
- `TabTransformer` koristi embeddings + self-attention i bolje iskorišćava bogate kategorije; nakon stabilizacije treninga postiže niži RMSE (~3.5k EUR).
- Imajte u vidu da je ovo **quick** režim (manje epoha, uži grid). Potpuni trening (duže trajanje) dodatno smanjuje grešku, ali je za potrebe prezentacije ovakav izlaz dovoljan i reproducibilan u realnom vremenu.

```python
# Cell 11: Ilustrativni feature importance (brzi fit na subsetu)
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import RobustScaler
# Prepare a compact training set for demonstration (fast)
cols_cat = ['Marka', 'Karoserija', 'Gorivo', 'Menjac']
cols_num = ['Godina_proizvodnje', 'Zapremina_motora', 'Kilometraza', 'Konjske_snage']
# Simple cleaning
demo = df.copy().dropna(subset=['Cena'])
# keep top 8 brands, others->Other
top = demo['Marka'].value_counts().nlargest(8).index
demo['Marka2'] = demo['Marka'].where(demo['Marka'].isin(top), 'Other')
X = pd.get_dummies(demo[['Marka2'] + cols_num], drop_first=True)
y = np.log1p(demo['Cena']).values
sc = RobustScaler()
num_cols_present = [c for c in cols_num if c in X.columns]
if num_cols_present:
    X[num_cols_present] = sc.fit_transform(X[num_cols_present])
# fit small GBR for demo
gbr = GradientBoostingRegressor(n_estimators=300, learning_rate=0.05, max_depth=4, random_state=42)
gbr.fit(X, y)
preds = gbr.predict(X)
rmse_demo = mean_squared_error(np.expm1(y), np.expm1(preds), squared=False)
print(f'Demo GBR RMSE on full data (no CV, demo only): {rmse_demo:.2f}')
# feature importances
fi = pd.Series(gbr.feature_importances_, index=X.columns).nlargest(12)
plt.figure(figsize=(8,4))
sns.barplot(x=fi.values, y=fi.index, palette='deep')
plt.title('Top feature importances (demo GBR)')
plt.show()
```

## Dodatna interpretacija — Feature Importances
Iako TabTransformer postiže nižu grešku, Gradient Boosting je lakše interpretirati. U nastavku pravimo brzi fit (bez kros-validacije) samo da bismo vizuelno pokazali koje karakteristike imaju najveći uticaj.
